```python
# Which team won the most matches?
match_winner = df['match_winner'].value_counts()
sns.barplot(y = match_winner.index , x = match_winner.values , palette='rainbow')
```

```python
# Toss Decision Trends
sns.countplot(x = df['toss_decision'])
```

```python
# Toss Winner vs Match Winner
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
persantage = (count * 100) / df.shape[0]
print(persantage.round(2))
```

```python
# How do teams win? (Runs vs Wickets)
df['won_by'].value_counts()
sns.countplot(x = df['won_by'])
```

```python
# Key Player Performances
```

```python
# Most "Player of the Match" Awards
count = df['player_of_the_match'].value_counts().head(5)
sns.barplot(y = count.index , x = count.values , palette = 'rainbow')
```

```python
# Top Scorers
count =df.groupby('top_scorer')['highscore'].sum().sort_values(ascending = False).head(10)
sns.barplot(x = count.values , y = count.index)
```

```python
# 10 Best Bowling Figures
df.head()
df['Higgeest_wickets'] = df['best_bowling_figure'].apply(lambda x : x.split('--')[0])
df['Higgeest_wickets'] = df['Higgeest_wickets'].astype(int)
count =df.groupby('best_bowling')['Higgeest_wickets'].sum().sort_values(ascending = False).head(10)
sns.barplot(x = count.values , y = count.index , palette = 'rainbow')
```

```python
# Custom Questions & Insights
```

```python
#  Who won the highest margin by runs?
df[df['won_by'] == 'Runs'].sort_values(by = 'margin' , ascending = False).head(1)[['match_winner' , 'margin']]
```

```python
# Which player had the highest individual score?
df[df['top_scorer'] == df['top_scorer']].max()[['top_scorer' , 'highscore']]
```

```python
# Which bowler had the best bowling figures?
df[df['Higgeest_wickets'] == df['Higgeest_wickets'].max()][['best_bowling' , 'best_bowling_figure']].head(1)
```
