```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Matplotlib Basic
x = np.linspace(0,11,10)
y = np.linspace(10,21,10)
plt.plot(x,y)
plt.plot(y,x)
plt.xlabel("Chacha")
plt.ylabel("Chachi")
plt.title("Hello")
plt.legend(["Chacha","Chachi"])
plt.show()
```

```python
#  Subplots in Matplotlib
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x**2,y)
plt.subplot(2,2,3)
plt.plot(x**3,y)
plt.subplot(2,2,4)
plt.plot(x,y**2)
```

```python
# Object Oriented way
fig = plt.figure(figsize=(8,6),dpi=100)
fig.add_axes([0.1,0.1,0.8,0.8])
plt.plot(x,y)
fig.add_axes([0.3,0.2,0.4,0.4])
plt.plot(x**2,y)
plt.show()
```

```python
# Types of Plots
plt.scatter(x,y)
```
