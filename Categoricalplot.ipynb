```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("tips")
```

```python
df
```

```python
# Countplot
```

```python
sns.countplot(x = df['sex'] , hue = df['smoker'])
```

```python
# Barplot
```

```python
sns.barplot(x = df['sex'] , y = df['smoker'])
```

```python
sns.barplot(x =  df['sex'], y = df['smoker'] ,estimator = np.sum)
```

```python
# Boxplot
```

```python
sns.boxplot(x = df['sex'] , y = df['total_bill'])
```

```python
sns.boxplot(x = df['day'] , y = df['total_bill'] , hue = df['smoker'])
```

```python
# Swarmplot
```

```python
sns.swarmplot(x = df['day'] , y = df['tip'])
```

```python
# Violen Plot
```

```python
sns.violinplot(x = df['day'] , y = df['total_bill'])
```

```python
sns.stripplot(x = df['day'] , y = df['total_bill'])
```
