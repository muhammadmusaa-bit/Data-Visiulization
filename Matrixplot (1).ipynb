```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
tips
```

```python
flights = sns.load_dataset('flights')
```

```python
flights
```

```python
# HeatMap
```

```python
flight = flights[['year' , 'passengers']]
flightcorr = flight.corr()
```

```python
sns.heatmap(flightcorr)
```

```python
sns.heatmap(flightcorr , annot = True)
```

```python
# ClusterMap
```

```python
sns.clustermap(flightcorr)
```

```python
# HeatMap with Piviot table
```

```python
pvt = flight.pivot_table(values='passengers' , index= 'year')
```

```python
pvt
```

```python
sns.heatmap(pvt , annot=True)
```

```python
pvt1 = tips.pivot_table(values='total_bill' , index='day' , columns='smoker')
```

```python
pvt1
```

```python
sns.heatmap(pvt1 , annot=True)
```
